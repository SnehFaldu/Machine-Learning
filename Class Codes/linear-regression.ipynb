```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv("train.csv")

df.shape
df.describe()
```

```python
y_box = y_value.boxplot(column=['y'])
plt.show()
```

```python
#ML modeling
def hypothesis(theta_array,x):
    return (theta_array[0]+theta_array[1]*x)

def cost_function(theta_array,x,y,m):
    total_error = 0
    for i in range(m):
        total_error = ((theta_array[0]+theta_array[1]*x[i])-y[i])**2
    return (total_error/(2*m))

def gradient_descent(theta_array,x,y,m,alpha):
    sum_0,sum_1 = 0,0
    for i in range(m):
        sum_0 += ((theta_array[0]+theta_array[1]*x[i])-y[i])
        sum_1 += x[i]*((theta_array[0]+theta_array[1]*x[i])-y[i])
    update_theta0 = theta_array[0] - (alpha/m)*sum_0
    update_theta1 = theta_array[1] - (alpha/m)*sum_1
    return (update_theta0,update_theta1)

def training(x,y,alpha,iterations):
    theta0,theta1 = 0,0
    theta_array = [theta0,theta1]
    m = x.size
    cost_value = []
    for i in range(iterations):
        theta_array = gradient_descent(theta_array,x,y,m,alpha)
        cost_value.append(cost_function(theta_array,x,y,m))
        y_new = theta_array[0]+theta_array[1]*x
        plt.plot(x,y_new,'r')
        plt.scatter(x,y)
        plt.show()
    x=np.arange(0,len(cost_value),step=1)
    plt.plot(x,cost_value)
    plt.show()
    return (theta_array,cost_value,y_new)
```

```python
clean_data = df.dropna().reset_index(drop=True)
clean_data.describe()

x_train = clean_data['x'].to_numpy()
y_train = clean_data['y'].to_numpy()
```

```python
alpha = 0.0001
iterations = 50
training(x_train,y_train,alpha,iterations)
```
